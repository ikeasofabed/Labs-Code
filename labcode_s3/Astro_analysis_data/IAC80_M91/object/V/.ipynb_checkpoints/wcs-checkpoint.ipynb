```python
from astropy.io import fits
from matplotlib import pyplot as plt
from astropy.wcs import WCS
from astropy.coordinates import SkyCoord
from photutils import DAOStarFinder
from astropy.stats import mad_std

%matplotlib inline
```

```python
#Read in and plot an image

image = fits.open("O130311_0123.fits")
plt.imshow(image[0].data, vmin=400, vmax=700)
plt.colorbar()
```

```python
# Now lets read in the WCS and see what this looks like
# note that the axes now show RA and Dec

image_wcs = WCS(image[0].header)
plt.subplot(projection=image_wcs)
plt.imshow(image[0].data, vmin=400, vmax=700)
plt.colorbar()
```

```python
# Find some sources (and put them in a table called "sources")

bkg_sigma = mad_std(image[0].data)
daofind = DAOStarFinder(fwhm=6., threshold=5.*bkg_sigma)
sources = daofind(image[0].data)

sources
```

```python
# Plot these sources on our image

plt.imshow(image[0].data, vmin=400, vmax=700)
plt.scatter(sources['xcentroid'], sources['ycentroid'], alpha=0.3, color="orange")
```

```python
# Now, for each source, get the RA and Dec

for source in sources:
    # We make a SkyCoord object from the pixel positions, using the world coordinate
    # information from our image (image_wcs).
    coords = SkyCoord.from_pixel(source['xcentroid'],source['ycentroid'],image_wcs,origin=0) 
    # We can easily access the RA and Dec of each source as follows
    ra = coords.ra.degree
    dec = coords.dec.degree
    
    print (ra, dec)
    
    # Somewhere here you could do
    # Open my Gaia catalog
    # Loop over Gaia catalog:
    #     if (ra - gaia_ra < distance) and (dec - gaia_dec < distance):
    #          print ("source match")
    #          get 1/parallax from catalog and calculate abs mag
```
