```python
cali_V = np.array([1, 1.25, 1.5, 1.75, 2, 2.25, 2.5, 2.75, 3])
cali_f = np.array([64.30, 80.29, 96.34, 112.14, 128.04, 144.03, 160.36, 176.50, 192.72])

plt.plot(cali_V, cali_f,'o')

from scipy.stats import linregress

slope, intercept, r_value, p_value, std_err = linregress(cali_V, cali_f)
print(f"The slope is: {slope}")
coefficients = np.polyfit(cali_V, cali_f, 1)                                    
y_intercept = coefficients[1]

print(y_intercept)

plt.show()
```

# Aluminium

```python
voltages = np.linspace(1, 4, 100)
```

```python
amp60 = []

for i in tqdm.tqdm(voltages):
    myDAC.writeVoltage(i)
    sleep(0.2)
    amp60.append(np.mean(myADC.sampleVoltages(30,100)[0]))
    
amp60 = np.asarray(amp60)
```

```python
plt.figure(figsize=(10,8))
plt.scatter(voltages, amp60, marker='.')
plt.grid()
plt.title('Aluminium at 60C pm 3')

plt.show()

print(amp60)
print(voltages)
```

```python
data = np.savetxt('aluminium 60C', (voltages, amp60))
```
