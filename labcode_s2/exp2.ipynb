```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt('/Users/JoanaUCD/Library/CloudStorage/OneDrive-UniversityCollegeDublin/Labs/labs code/exp2data.txt')

x = data[:,0] # 2 delta
y = data[:,1] # xray counts

errAngle = 0.5
errCount = np.sqrt(y)
```

```python
plt.figure(figsize=(15,10))

plt.errorbar(x, y, errCount, errAngle, fmt=",", ecolor="#f5b8e3", color="#ff30c4", capsize=4, barsabove=True, zorder=3, label="Errorbars")

plt.xlabel(r"$2\theta$")
plt.ylabel("X-Ray Count")

plt.plot(x, y, color="#ff30c4", zorder=4)
plt.scatter(x,y, color="#bf088b", s=7, zorder=5, label="Readings")

plt.minorticks_on()
plt.grid(True, which="major", linewidth=0.8, color="#DDDDDD", zorder=2)
plt.grid(True, which="minor", linewidth=0.5, color="#EEEEEE", linestyle="--", zorder=1)

plt.title("Graph of Count against Angle for the LiF Crystal at 30kV")

plt.legend(fontsize="xx-large")

plt.show()
```
