## Calculations

```python
V = 386
T = 28.8
d = 0.00767
rho = 886
eta = 1.867*10**-5
P = 1.01*10**5
b = 8.2*10**-3
g = 9.8

e = -1.6*10**-19
```

```python
# Electric Field
E = V/d

print(E, "V/m")
```

```python
# average radius of the oil drop
a = (np.sqrt((b/(2*P))**2+((9*eta*vf2)/(2*rho*g)))-b/(2*P))

print(a, "m")
print(np.average(a), "m")
```

```python
# weight of the drop
m = (4/3)*np.pi*((np.average(a))**3)*rho*g

print(m, "N")
```

```python
vf3_compatible = vf3[0:11]
print(vf3_compatible, np.size(vf3_compatible))

# charge
q = (4*np.pi)/(3) * (np.sqrt((((b)/(2*P))**2)+((9*eta*vf4)/(2*rho*g)))-(b)/(2*P))**3 * (rho*g*d*(vf4-vf3_compatible))/(V*vf4)

print(q, np.size(q))

n = q/e
print(n, np.size(n))
```

## Uncertainties

```python
dist_err = 0.00005
time_err = 0.005
# Time and Distance errors for all sample values

ourtime_err = 0.5
# Only because we did it to full seconds + minutes rather than .seconds
```

```python
# uncertainty on radius a
un_a = np.std(a)/(np.sqrt(np.size(a)))

print(un_a)
```

```python
q_err = np.std(q)/(np.sqrt(np.size(q)))
print(q_err)
```
